## §1 What we measure — an operational definition of tacit knowledge

**Object of study.** For a metric ω (an R3-level evaluative concept, e.g. *"Ethical target selection
(punching up)"*) and an executor class E (here Llama-3.1-8B as the reference executor), the executor's
*knowing* is its reproducible judgment pattern: the average judgment pattern across different wordings of the rubric
$\bar M_\omega = \mathbb{E}_{\varphi}[M_\omega^{\varphi}]$ (we try 4 different phrasings and average — this removes luck from which exact words we chose) (4 forms), evaluated on 300 probe texts. Averaging over the orbit removes
single-wording luck from the **target seat**.

**Tacit knowledge, operationally** (Polanyi's "we know more than we can tell", made measurable):
content of $\bar M_\omega$ that is *replicable* (so it is knowledge, not noise) but *not carried by any
articulated checklist we can certify*. Two independent measurements triangulate it:

*Terminology note (construct validity).* "The knowing" is used strictly in the
operational sense of **replicable judgment structure**. Whether that structure tracks the
intended construct — the quality dimension — rather than executor idiosyncrasy is a
construct-validity question in Cronbach's sense, bounded separately: the orbit/forms are
generalizability facets (reliability), own-vs-cross metric correlations give discriminant
evidence (identity), the dense ceiling (our strongest baseline: a kitchen-sink feature model with no rubric at all) gives convergent evidence, and the human bracket is
the eventual criterion. A Δ > 0 against an invalid target would still demonstrate
*inarticulable structure*, but not *inarticulable knowledge* — which is why the validity
checks ride along with every claim.

**Face 1 — the census ceiling.** From the articulated criteria pool Ω (species = behavioral
equivalence classes of criteria; paraphrases are quotiented together), build the best explicit
checklist: $\mathrm{OPT}_\Omega$ = bits of $\bar M_\omega$ recovered by the optimal criteria head.
Capture–recapture missing-mass machinery certifies an ε such that *no unseen articulable species worth
more than ε is hiding* (anytime-valid; §5). Then
$$\Delta(E) \;=\; \underbrace{H(\bar M_\omega)}_{\text{the knowing}} \;-\; \underbrace{[\mathrm{OPT}_\Omega + \varepsilon]}_{\text{certified telling ceiling}}\;>0$$
is a **certified tacit residual** — with the standing bracket discipline: recovery numbers are LOWER
bounds on the ideal articulable content, census $B_E$ is the UPPER side; the two never collapse.

**The form gate (pre-census).** Before asking *how much* a checklist captures, each criterion must
survive re-articulation: a criterion is form-invariant iff its median binary flip rate under
meaning-preserving paraphrase is ≤ 10% (measured on the SAME executor, same paraphrase generator, vs a
repeat-noise floor τ₀). **FORM-DOMINATED** = the telling won't hold still in words — a *pre-tacit*
failure: you cannot census what you cannot stabilize. This gate fires on the **instrument seat**
(criteria wording), which orbit-averaging of the target does not touch.

**Face 2 — decompression.** Fix the reference judgment $\bar M_\omega$ (8B executor). Write messages of
increasing articulation TYPE — `name → definition → explanation → full_rubric → exemplars → dossier`
(definition/explanation are length-matched ~25–40 words, written by a 70B writer that NEVER sees
labels — so this measures *transmission*, not supervised fitting). Readers of increasing capacity
(1B / 3B / 8B / soon 70B) reconstruct the judgment from each message alone.
$R_E(\text{rung}, \text{reader})$ = balanced accuracy vs the reference. The **reader gap**
$G(r) = R(\text{strong}) - R(\text{weak})$ locates where the knowledge is packed: a gap at the *name*
rung means the name is a compressed pointer into background the strong reader already has; the gap
closing as messages get richer shows knowledge being moved INTO the channel. `span_R²` asks whether a
rung's judgment is an assembly of already-articulated species (in-span) or something new (out-of-span
→ candidate *ostension*).

**Three different "deltas" appear in this work — do not conflate them:**
1. **Face-2 reader gaps** (e.g. +.104): balanced-accuracy differences 3B−1B at a rung. Behavioral,
   per-message-type.
2. **Δ(E), the certified residual** (bits): $H(\bar M) - [\mathrm{OPT}_\Omega + \varepsilon]$ — Face-1's
   tacitness bound (provisional until adversarial saturation confirms; §5).
3. **`delta` in the certificate rows**: the confidence parameter δ of the anytime-valid guarantee
   (a union-bound budget, not a knowledge quantity).

## §2 Face 1 — census-ceiling certificates

Per metric, the certificate reports: target entropy `H_M` (how many bits of yes/no information the metric carries — more bits = more decisive verdicts),
`opt_omega_bits` (the certified checklist head OPT_Ω), `frac_H` = OPT_Ω/H (head share), `eps_bits`
(the certified missing-mass band ε), the per-criterion marginal `gains`, the form-gate flag, and the
verdict: **CODIFIABLE** (band closed — no room for a tacit residual) / **UNDERSAMPLED** (band still
open — a *sampling* problem, curable with more probes) / **FORM-DOMINATED** (blocked at the form gate
— words don't hold still, census not interpretable). D1 filter: metrics with `H_M < 0.15` bits are
near-constant readouts — vacuously certifiable, excluded.

```python
rows, AUD = [], {}
for dom in DOMS:
    cert = CERT[dom]
    a = audit_cert(cert, floor=0.15); AUD[dom] = a
    kept = [r for r in cert if (r.get("H_M") or 0) >= 0.15]
    med = lambda k: float(np.median([r[k] for r in kept if r.get(k) is not None]))
    rows.append({"domain": dom,
                 "n metrics": a["n_total"], "degenerate": a["n_degenerate"], "kept": a["n_keep"],
                 "FORM-DOM": a["verdicts_kept"].get("FORM-DOMINATED", 0),
                 "UNDERSAMP": a["verdicts_kept"].get("UNDERSAMPLED", 0),
                 "CODIFIABLE": a["verdicts_kept"].get("CODIFIABLE", 0),
                 "form-gate PASS": f"{a['form_gate_pass_kept']}/{a['n_keep']} = {a['form_gate_pass_rate']:.0%}",
                 "med OPT_Ω/H": round(med("frac_H"), 3),
                 "med ε (bits)": round(med("eps_bits"), 3),
                 "med H_M (bits)": round(med("H_M"), 3)})
    if a["spurious_codifiable"]:
        print(f"[{dom}] D1 caught spurious CODIFIABLE on degenerate target: {a['spurious_codifiable']}")
face1 = pd.DataFrame(rows).set_index("domain")
face1
```

```python
fig, axes = plt.subplots(2, 2, figsize=(11.5, 7.5))

# (a) verdict mix (kept)
ax = axes[0, 0]
verds = ["FORM-DOMINATED", "UNDERSAMPLED", "CODIFIABLE"]
x = np.arange(len(verds)); w = 0.36
for i, dom in enumerate(DOMS):
    n = AUD[dom]["n_keep"]
    vals = [AUD[dom]["verdicts_kept"].get(v, 0) / n for v in verds]
    ax.bar(x + (i - 0.5) * w, vals, w, label=f"{dom} (n={n})", color=DCOL[dom])
ax.set_xticks(x); ax.set_xticklabels(["FORM-\nDOMINATED", "UNDER-\nSAMPLED", "CODIFIABLE"])
ax.set_ylabel("fraction of kept metrics"); ax.set_title("Face-1 verdict mix (D1-filtered)"); ax.legend()

# (b) form-gate pass rate
ax = axes[0, 1]
for i, dom in enumerate(DOMS):
    ax.bar(i, AUD[dom]["form_gate_pass_rate"], 0.55, color=DCOL[dom])
    ax.text(i, AUD[dom]["form_gate_pass_rate"] + 0.01,
            f"{AUD[dom]['form_gate_pass_kept']}/{AUD[dom]['n_keep']}", ha="center")
ax.set_xticks(range(len(DOMS))); ax.set_xticklabels(DOMS)
ax.set_ylabel("criteria form-gate pass rate")
ax.set_title("Form-invariance gate (≤10% median flip under paraphrase)\nsame executor + same paraphrase instrument in both domains")

# (c) head-share ECDF
ax = axes[1, 0]
for dom in DOMS:
    v = np.sort([r["frac_H"] for r in CERT[dom] if (r.get("H_M") or 0) >= 0.15 and r.get("frac_H") is not None])
    ax.plot(v, np.arange(1, len(v) + 1) / len(v), label=dom, color=DCOL[dom])
ax.set_xlabel("OPT_Ω / H  (share of judgment entropy captured by the certified checklist head)")
ax.set_ylabel("ECDF"); ax.set_title("Heads are strong in both domains"); ax.legend()

# (d) diminishing returns of articulation
ax = axes[1, 1]
K = 20
for dom in DOMS:
    curves = []
    for r in CERT[dom]:
        if (r.get("H_M") or 0) < 0.15 or not r.get("gains"):
            continue
        c = np.cumsum(r["gains"])[:K] / r["H_M"]
        if len(c) < K:
            c = np.pad(c, (0, K - len(c)), constant_values=c[-1])
        curves.append(c)
    med = np.median(np.vstack(curves), axis=0)
    ax.plot(range(1, K + 1), med, marker="o", ms=3, label=dom, color=DCOL[dom])
ax.set_xlabel("k-th criterion added to the checklist head")
ax.set_ylabel("median cumulative OPT_Ω / H")
ax.set_title("Diminishing returns of articulation\n(first ~5 criteria carry most of what words can carry)")
ax.legend()
plt.tight_layout(); plt.show()
```

```python
# Provisional certified residual (the Δ(E) direction) — ONLY meaningful where the form gate passes.
recs = []
for dom in DOMS:
    for r in CERT[dom]:
        if (r.get("H_M") or 0) >= 0.15 and r.get("form_invariant") is True:
            resid = r["H_M"] - (r["opt_omega_bits"] + (r.get("eps_bits") or 0))
            recs.append({"domain": dom, "metric": r["name"][:44], "verdict": r["verdict"],
                         "H_M": r["H_M"], "OPT_Ω": r["opt_omega_bits"], "ε": r.get("eps_bits"),
                         "residual_bits": resid, "adv_saturated": r.get("adv_saturated")})
resid = pd.DataFrame(recs).sort_values(["domain", "residual_bits"], ascending=[True, False])

fig, axes = plt.subplots(1, 2, figsize=(12.5, 5.5), sharex=True)
for ax, dom in zip(axes, DOMS):
    sub = resid[resid.domain == dom].sort_values("residual_bits")
    cols = ["tab:green" if v == "CODIFIABLE" else "tab:gray" for v in sub.verdict]
    ax.barh(sub.metric, sub.residual_bits, color=cols)
    ax.axvline(0, color="k", lw=0.8)
    ax.set_title(f"{dom} — form-gate passers (n={len(sub)})")
    ax.set_xlabel("H_M − (OPT_Ω + ε)   [bits]  — PROVISIONAL residual")
fig.suptitle("Provisional tacit residual per gate-passing metric (green = CODIFIABLE ⇒ band closed, residual ≈ ≤0)\n"
             "UNDERSAMPLED bars are an OPEN ε-band, not yet a certified tacit claim (adv_saturated pending)", y=1.04)
plt.tight_layout(); plt.show()
resid.round(3).reset_index(drop=True)
```

**Reading Face 1.**

**How to read these plots.** Each panel shows a different view of the certificate measurements:
- **(a) Top-left**: Verdict mix — what share of metrics fall into each outcome category (CODIFIABLE / UNDERSAMPLED / FORM-DOMINATED)
- **(b) Top-right**: Form-gate pass rate — what fraction of metrics have stable verdicts across different wordings
- **(c) Bottom-left**: Checklist coverage — for each metric, how much of its information (H_M) the best checklist recovered (OPT_Ω / H_M)
- **(d) Bottom-right**: Missing-mass band — the ε estimate (how much more we might find with better sampling)

**Takeaway:** Form fragility (whether verdicts hold still across rewordings) is domain-specific — humor metrics are more stable (35% pass) than creative writing (16% pass).

- **Form fragility is domain-specific, not universal** — gate pass 35% (humor) vs 16% (CW) on the
  *same* executor and the *same* paraphrase instrument. Since the executor's generic paraphrase
  sensitivity is held constant, the difference is a property of the *criteria content*: humor's
  evaluative concepts (setup/incongruity/target/timing) pin stable decision functions that survive
  rewording; CW's wordings are doing more indexical work. (Stable 41→60 metrics: 36%→35%.)
- **Heads are strong in both** (median OPT_Ω/H ≈ 0.69–0.74): most of what CAN be captured is captured
  by the first handful of articulated criteria — value is head-concentrated, arguing against
  deep-tail tacitness in both domains.
- **Humor reaches value verdicts more often** (15 US + 4 COD vs 5 + 2): fewer metrics are blocked at
  the gate, so the census actually gets to run; its open bands are a sampling problem, curable.
- **CODIFIABLE is provisional**: `adv_saturated` is not yet confirmed on these runs (D3) — the ε-bound
  alone cannot exclude adversarially-hidden tail synergies (the tail-XOR construction hides 0.096 bits
  at ε=.031 — a measured 3.1× under-cover; §5).

```python
def med_cell(rep, sub, rung, key="bal_acc"):
    vals = [g[rung][key] for tag, gs in rep.items() if sub in tag for g in gs.values() if rung in g and key in g[rung]]
    return float(np.median(vals)) if vals else np.nan

def per_metric(rep, sub, rung, key="bal_acc"):
    for tag, gs in rep.items():
        if sub in tag:
            return {gi: g[rung][key] for gi, g in gs.items() if rung in g}
    return {}

tabs = {}
for dom in DOMS:
    rep = GRID[dom]
    t = pd.DataFrame({rd: [med_cell(rep, rd, r) for r in RUNGS] for rd in ["1B", "3B", "8B"]}, index=RUNGS)
    # canonical clean gap = median over metrics of per-metric (3B − 1B)
    e3, e1 = {r: per_metric(rep, "3B", r) for r in RUNGS}, {r: per_metric(rep, "1B", r) for r in RUNGS}
    t["3B−1B (clean gap)"] = [float(np.median([e3[r][gi] - e1[r][gi] for gi in e3[r] if gi in e1[r]])) for r in RUNGS]
    t = t.rename(columns={"8B": "8B (REFERENCE — self-ref, excluded from gaps)"})
    tabs[dom] = t.round(3)
    print(f"\n=== {dom} — median balanced accuracy per rung ===")
    display(tabs[dom])
```

### §4c Expansion-chain v1 results (landed 2026-07-03 01:24) — and what the planted controls caught

**How to read the expansion results.** The planted controls are metrics we deliberately corrupted — if the measurement is working, these should show detectably worse performance. The "chain" measurements show which types of added detail (definitions vs examples vs background knowledge) actually improve executor performance.


The nested-chain experiment ran end-to-end overnight (38 metrics + 4 planted controls × 8 levels ×
1B/3B/70B readers). **The planted controls did their job — they caught two real confounds** that
change how the results must be read:

1. **The 1B is not a valid instrument.** It scores at *chance* on `planted_question` ("does the
   text contain a question mark?") in both domains, flat across all 8 levels, while producing
   highly stable verdicts (self_agree ≈ .97). Stable-but-flat = it cannot execute ANY judgment
   instruction, mechanical or conceptual. Therefore 1B-censoring cannot distinguish tacit content
   from generic incapacity: the 1B is a demonstrative *floor*, not a graded reader. The graded
   iso-performance story lives in **3B → 70B**.
2. **Gold-vs-view truncation artifact.** Planted gold was computed on FULL probe texts; readers see
   a 4000-char truncated view (`max_text_chars`). Tail-sensitive rules are therefore capped for
   long-text CW (70B question-detection .66–.72) but nearly fine for short-joke humor (.81–.91) —
   the exact domain asymmetry observed. Fix for v2: compute planted gold on the truncated view.
   (`planted_length150` was additionally gold-imbalanced (4% minority) — flagged automatically.)
3. **A 3B compliance ceiling ≈ .7–.75 on mechanical rules** — so "censored at 3B" claims on metrics
   whose 70B bar exceeds ~.75 may be compliance-capped rather than content-capped. Noted per item.

## §5 The guarantees — what "certified" means here

The theory home is `notes/2026-07-02__two-faces-theory.md` +
`notes/2026-06-18__prompt-optimality-theory.md` §12.8; this is the working summary.

**The bracket.** $R(p) \le T(m_\omega) \le I(M^*;X) \le U_{B_E}$. Every recovery number (Face 2, and
OPT_Ω itself) is a **lower bound** on ideal articulable content — a better writer/reader could do
better. The census side ($B_E$, species counting) gives the **upper** flank. They must never be
collapsed into one number; tacitness claims live in the gap between the *certified upper ceiling of
telling* and the *replicable knowing*.

**Face-1 certificate machinery** (implemented in `value_certificate.py`, run per metric):
1. **T1 — anytime-valid ε**: the missing-mass band `eps_bits` is valid under optional stopping
   (δ-union over checkpoints × orders with an a-priori B-cap). δ appears as `delta` in the rows.
2. **I1 — OSW horizon**: extrapolation of the species curve is clamped at $c \le \ln N$ — no
   scaling-law wishful thinking beyond the observation window.
3. **C2 — the adversarial flank**: a constructed pairwise-independent parity (tail-XOR) triple shows a
   metric can be declared CODIFIABLE at ε=.031 while 0.096 bits hide in criteria *synergies* — a
   measured **3.1× under-cover**. The `adv_saturated` gate (adversarial generator arms fail to find
   new value) is the mitigation. **In today's runs `adv_saturated=None` ⇒ all CODIFIABLE verdicts are
   PROVISIONAL (audit D3).**
4. **Form gate**: median criterion flip ≤ 10% under meaning-preserving transformations (vs repeat-noise
   floor τ₀). The orbit's transformations are largely NON-CONTENT presentational edits — boilerplate
   wrap, statement→question, clause reorder, suffix append — i.e. exactly the "shouldn't matter"
   perturbations open models are known to be brittle to. The judge's OUTPUT protocol (free-text vs
   JSON, chat template) is held constant everywhere — measured elsewhere, not varied here. ~52% of raw form effect is a calibratable strictness main-effect (removable); the gate
   fires on the residual item×form entanglement. FORM-DOMINATED metrics never reach a value verdict.
5. **Audit disciplines** (`audit_certificate.py`, tested): **D1** drop `H_M < 0.15` bits (vacuous
   targets — caught 1 spurious CODIFIABLE in humor twice, at 41 and at 60); **D2** exclude the
   reference executor from reader gaps (an earlier 8B-inclusive "+.245" was retracted); **D3**
   CODIFIABLE requires saturation.

**Face-2 honesty conditions.** The writer never sees labels (transmission, not fitting — and not
GEPA: no optimization loop against recovery). Definition/explanation are length-matched, so type is
not confounded with length (length alone was already tested and is a poor predictor). Exemplar probes
are excluded from evaluation. Verbal rungs are averaged over 3 paraphrase forms.

**Direction-of-error discipline** (the anthropological framing): weak instruments *inflate* apparent
tacitness — an incapable writer/reader makes everything look untellable. Credibility therefore rests
on positive controls, which this run supplies: CODIFIABLE metrics exist (band closes when content is
truly checklistable), humor exemplars transmit (the ostension channel works when content is showable),
and heads are strong (the census machinery finds real value where it exists).

**Validity framing (Cronbach).** The reliability side (τ₀ repeat floors, the form orbit as
generalizability facets) is necessary but not sufficient; construct validity enters through
convergent (dense ceiling), discriminant (own-vs-cross metric identity), and — planned —
criterion (human-arm) evidence. All measurements here are of *transmission of the target as
operationalized*.

**Current scope limits.** Single executor family (Llama) with the 8B as reference; single proposer
family (GLM-4.7) for Ω; 70B-as-reader queued (real dynamic range); native 70B/Qwen criteria
signatures pending the Q1/Q2/Q3 fork (the `aligned_70b_orbit_v2` / `aligned_qwen_orbit_v2` dirs are

```python
# §6b.1 gi35 sealed result — calibration vs sealed, all six substantive arms
V2 = DATA / "concluding_policy_confirmation_v2"

def _arm_rows(report_path):
    r = json.load(open(report_path))
    out = {}
    for cell in r["cells"]:
        for row in (cell.get("arms") or cell.get("rows")):
            if row.get("control_for") is not None:
                continue
            sim = row.get("scale_step_specificity_simultaneous_certificate")
            if not isinstance(sim, dict):
                continue
            tf = sim["target_fidelity"]
            gap = row["scale_step_certificate"]["differences"]["native_rho_larger_minus_small"]
            out[(cell["cell_id"], row["arm_id"])] = dict(
                adverse=tf["candidate_adverse_rho_point"], adverse_CI=tf["candidate_adverse_rho_CI"],
                quotient=tf["candidate_quotient_rho_point"], quotient_CI=tf["candidate_quotient_rho_CI"],
                tier=tf["certified"]["tier"], native_gap=gap["point"])
    return r, out

lock_r, lock = _arm_rows(V2 / "lockbox_report.json")
cal_r, cal = _arm_rows(V2 / "calibration_report.json")
assert lock_r["partition"] == "tacit_breadth_validation" and cal_r["partition"] == "tacit_breadth_search"
print("sealed simultaneous-certified joint substitutions:",
      lock_r["summary"]["n_simultaneous_certified_joint_fixed_target_endpoint_isomorphic_scale_substitutions"], "of 3 prior-selected constructs\n")
rows = []
for (cell, arm), v in sorted(lock.items()):
    rows.append({"construct": cell, "arm": arm,
                 "calibration adverse": round(cal[(cell, arm)]["adverse"], 3),
                 "SEALED adverse": round(v["adverse"], 3),
                 "sealed CI": [round(x, 3) for x in v["adverse_CI"]],
                 "sealed quotient": round(v["quotient"], 3),
                 "native 70B-8B gap": round(v["native_gap"], 3),
                 "tier": v["tier"]})
pd.DataFrame(rows)
```

```python
# §6b.2 capacity ladder — definition arm, adverse rho vs the frozen 70B name target
# EXPLORATORY: 1B/3B are Llama-3.2 (cross-version); 8B is same-version, same open partition.
def _defn_adverse(path):
    r = json.load(open(path))
    out = {}
    for cell in r["cells"]:
        for row in (cell.get("arms") or cell.get("rows")):
            if row.get("control_for") is None and row["arm_id"] == "source_definition":
                tf = row["scale_step_specificity_simultaneous_certificate"]["target_fidelity"]
                out[cell["cell_id"]] = (tf["candidate_adverse_rho_point"], tf["candidate_adverse_rho_CI"])
    return out

RUNGS = [
    ("1B (3.2)", DATA / "concluding_policy_capacity_1b_exploratory_v1/calibration_report.json"),
    ("3B (3.2)", DATA / "concluding_policy_capacity_3b_exploratory_v1/calibration_report.json"),
    ("8B (3.1)", DATA / "concluding_policy_confirmation_v2/calibration_report.json"),
]
ladder = {label: _defn_adverse(path) for label, path in RUNGS}
labels = [label for label, _ in RUNGS]
fig, ax = plt.subplots(figsize=(6.5, 4))
for cell_id, marker in (("N_press-releases_35", "o"), ("N_humor_23", "s"), ("N_humor_11", "^")):
    pts = [ladder[l][cell_id][0] for l in labels]
    los = [ladder[l][cell_id][0] - ladder[l][cell_id][1][0] for l in labels]
    his = [ladder[l][cell_id][1][1] - ladder[l][cell_id][0] for l in labels]
    ax.errorbar(range(3), pts, yerr=[los, his], marker=marker, capsize=3, label=cell_id)
ax.axhline(.70, color="gray", ls="--", lw=1)
ax.set_xticks(range(3), labels)
ax.set_ylabel("definition-arm adverse rho vs frozen 70B name target")
ax.set_title("Capacity ladder (cross-version EXPLORATORY; open partition)")
ax.legend(fontsize=8)
fig.tight_layout()
```

```python
# §6b.3 interim unselected-panel tallies — one row per completed breadth domain
# Re-running this cell picks up new domains automatically as the executor lanes finish them
# (sync from sk3:~/interim_tallies/). Point estimates only; see claim boundary above.
rows = []
for path in sorted((DATA / "interim_tallies").glob("interim_tally_*.json")):
    drows = json.load(open(path))["rows"]
    domain = path.stem.replace("interim_tally_", "")
    gains = [r["gain_over_name"] for r in drows]
    cross = [r for r in drows if r["best_adverse"] >= .70]
    rows.append({"domain": domain, "n_cells": len(drows),
                 "median native gap": round(np.median([r["native_gap_point"] for r in drows]), 3),
                 "median articulation gain": round(np.median(gains), 3),
                 "cells w/ negative gain": sum(1 for g in gains if g < 0),
                 "best-arm >= .70": f"{len(cross)}/{len(drows)}",
                 "...and beats controls": sum(1 for r in cross if r["best_beats_its_controls"])})
display(pd.DataFrame(rows))
lvl_rows = []
for path in sorted((DATA / "interim_tallies").glob("interim_tally_*.json")):
    drows = json.load(open(path))["rows"]
    for lvl in ("R1", "R2", "R3"):
        g = [r for r in drows if r["level"] == lvl]
        lvl_rows.append({"domain": path.stem.replace("interim_tally_", ""), "level": lvl,
                         "rescued": f"{sum(1 for r in g if r['best_adverse'] >= .70)}/{len(g)}"})
pd.DataFrame(lvl_rows).pivot(index="domain", columns="level", values="rescued")
```
